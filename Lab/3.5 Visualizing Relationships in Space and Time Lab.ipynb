```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the Dataset (Directly from the Seaborn library source)
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/healthexp.csv'
df = pd.read_csv(url)

print("✅ Data Loaded! Here is the Spatial and Time-Series data:")
print(df.head())

# Set the theme
sns.set_theme(style="whitegrid")

# --- VISUALIZATION 1: CHANGE OVER TIME (Time-Series) ---
# Question: How has Life Expectancy changed over the years?
plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='Life_Expectancy', hue='Country', data=df, marker='o')
plt.title('Viz 1: Life Expectancy Over Time (1970-2020)', fontsize=15)
plt.xlabel('Year')
plt.ylabel('Life Expectancy (Years)')
plt.show()

# --- VISUALIZATION 2: SPATIAL COMPARISON (Space) ---
# Question: Which country spends the most on average?
plt.figure(figsize=(10, 6))
avg_spend = df.groupby('Country')['Spending_USD'].mean().sort_values()
sns.barplot(x=avg_spend.index, y=avg_spend.values, palette='viridis')
plt.title('Viz 2: Average Healthcare Spending by Country (Spatial)', fontsize=15)
plt.ylabel('Average Spending (USD)')
plt.show()

# --- VISUALIZATION 3: SPACE vs TIME (Trajectory) ---
# Question: Does spending more money actually buy a longer life?
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Spending_USD', y='Life_Expectancy', hue='Country', size='Year', sizes=(20, 200), alpha=0.7, data=df)
plt.title('Viz 3: The Cost of Life (Spending vs. Life Expectancy)', fontsize=15)
plt.xlabel('Healthcare Spending (USD)')
plt.ylabel('Life Expectancy (Years)')
plt.show()
```

1. Dataset Description

"I selected the 'Health Expectancy' dataset, which tracks healthcare economics from 1970 to 2020. The data was sourced from the OECD (Organisation for Economic Co-operation and Development). It contains spatial features (Country), time-series features (Year), and quantitative metrics (Spending_USD and Life_Expectancy)."

2. Explanation for Viz 1 (Line Chart)

"This visualization tracks Life Expectancy over time. The feature changes consistently, showing a steady upward trend for all countries over the 50-year period, although the gap between countries like Japan and the USA has widened over time."

3. Explanation for Viz 2 (Bar Chart)

"This visualization compares the spatial component (Country) against spending. It reveals a massive disparity in space: the USA spends significantly more on healthcare than Germany or France, despite having similar or lower life expectancies."

4. Explanation for Viz 3 (Scatter Plot)

"This visualization combines space and time to show the trajectory of development. As time moves forward (represented by the size of the dots), all countries move 'up and to the right,' but the USA (Orange) moves further right (more expensive) without moving as far up (living longer) compared to other nations."
